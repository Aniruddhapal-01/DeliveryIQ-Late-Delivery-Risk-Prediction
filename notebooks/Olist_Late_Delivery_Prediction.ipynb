```python
customer_lat = np.radians(df_master['customer_lat'])
customer_lng = np.radians(df_master['customer_lng'])

seller_lat = np.radians(df_master['avg_seller_lat'])
seller_lng = np.radians(df_master['avg_seller_lng'])

delta_lat = seller_lat - customer_lat
delta_lng = seller_lng - customer_lng

a = (
    np.sin(delta_lat / 2)**2
    + np.cos(customer_lat) * np.cos(seller_lat)
    * np.sin(delta_lng / 2)**2
)

a = np.clip(a, 0, 1)

distance_km = 2 * 6371 * np.arcsin(np.sqrt(a))

df_master['customer_seller_distance_km'] = distance_km
```
